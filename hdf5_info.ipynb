```python
import h5py
hdf5_path = "/home/carl_lab/ola/stove_pot_bellpepper_lid_100/demo.hdf5"
```

```python
with h5py.File(hdf5_path, 'r') as f:
    def print_attrs(name, obj):
        print(name)
        for key, val in obj.attrs.items():
            print(f"    {key}: {val}")

    f.visititems(print_attrs)
```
